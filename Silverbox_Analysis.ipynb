# Silverbox coefficient analysis and quantum regression

Keep this notebook and `silverbox_models.py` in the same folder. Install dependencies in your Python environment: `pip install numpy pandas matplotlib scipy threadpoolctl`.


```python
import pandas as pd
from silverbox_models import CoefficientLeastSquares, QuantumSilverbox
```

## Load the measurements
Use existing CSV containing `V1` and `V2`.


```python
df = pd.read_csv("silverbox.csv")
df.head()
```

## Independent coefficient analysis
The default fit uses 1,000 randomly selected valid derivative rows. This exploratory fit is separate from QML calibration, which uses only revealed training outputs.


```python
coefficients = CoefficientLeastSquares(df)
X = coefficients.fit()
```

```python
coefficients.plot()
```

## Select the QML data segment
`start` is the zero-based position in the original dataframe; 

`count` is the number of consecutive observations. 
Both are required. The model preserves the supplied 60/20/20 chronological split and 96-input history.

You may also use a for loop to get several segmenst at once, the output are saved in pngs,and pdfs files in the same folder


## Run the limited-label comparison
`plot_count=150` is the default display length, capped at the available test prediction points. It changes only the trajectory plots, not the data or metrics. The original plotting code saves PNG/PDF plots in the working folder.


## Access all test predictions
Each dictionary key is `(fraction, seed)`.




```python
start = 80000
count = 5000
qml = QuantumSilverbox(df, start=start, count=count)
results, validation, summary = qml.run_limited_label_experiment(
    fractions=(0.02, 0.05, 0.08, 0.10, 0.25, 1.0),
    seeds=(50,),
    steps=10,
    plot_count=150
)
display(results)
display(validation)
display(summary)
display(qml.predictions[(0.02, 50)].head())
```
